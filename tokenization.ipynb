## parter Stremmer

```python
from nltk.stem import PorterStemmer
s=PorterStemmer()
```

```python
words=['playing','played','plays','study','studied']

for word in words:
    print("Root word=",word,"-->",s.stem(word))
```

```python
from nltk.stem import RegexpStemmer
reg=RegexpStemmer('ing$|s$|e$|able$|ed$', min=4)
```

```python
for word in words:
    print("Root word=",word,"-->",reg.stem(word))
```

```python
from nltk import pos_tag
words = word_tokenize(corpus)
print("Words:")
print(words)

# POS Tagging
pos = pos_tag(words)

print("POS Tagging:")
for word, tag in pos:
    print(f"{word} : {tag}")
```

## morphological analysis

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ["playing", "walked", "boys", "reading","plays","play"]

print("Word\tRoot\tSuffix")

for word in words:
    root = ps.stem(word)

    if word.startswith(root):
        suffix = word[len(root):]
    else:
        suffix = "Unknown"

    print(word, "\t", root, "\t", suffix)
```
